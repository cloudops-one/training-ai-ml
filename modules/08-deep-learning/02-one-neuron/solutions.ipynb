```python
# --- data helper: works locally and on Colab -------------------------------
from pathlib import Path
import urllib.request

def data_path(name):
    """Return a local path to data/<name>, downloading from GitHub on Colab."""
    for parent in [Path.cwd(), *Path.cwd().parents]:
        candidate = parent / "data" / name
        if candidate.exists():
            return candidate
    url = f"https://raw.githubusercontent.com/cloudops-one/training-ai-ml/main/data/{name}"
    dest = Path(name)
    if not dest.exists():
        urllib.request.urlretrieve(url, dest)
    return dest

import os
os.environ["TF_CPP_MIN_LOG_LEVEL"] = "2"        # hide TensorFlow start-up messages
import numpy as np, pandas as pd
import matplotlib.pyplot as plt
import warnings; warnings.filterwarnings("ignore")

x = np.array([1.5, -0.5, -0.8])      # C042: days_since_last, orders, spend (scaled)
w = np.array([0.9, -1.1, -0.4])      # weights
b = -0.3                             # bias

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

X = np.array([[ 1.5, -0.5, -0.8],    # C042
              [-0.7,  1.2,  0.9],    # a loyal customer
              [ 0.2,  0.0, -0.1]])   # an average customer
W = np.array([[ 0.9, -0.2,  0.5,  0.1],
              [-1.1,  0.4,  0.3, -0.6],
              [-0.4,  0.7, -0.2,  0.8]])
bias = np.array([-0.3, 0.1, 0.0, 0.2])
```

## Exercise 1

$w \cdot x = 0.9(-1.0) + (-1.1)(2.0) + (-0.4)(1.5) = -0.9 - 2.2 - 0.6 = -3.7$. Add $b$: $z = -4.0$. $\sigma(-4.0) = 1/(1 + e^{4}) = 0.018$.

A recent, frequent, high-spending customer: under 2% churn risk.

```python
x77 = np.array([-1.0, 2.0, 1.5])
z77 = x77 @ w + b
print(f"z = {z77:.2f}, churn probability = {sigmoid(z77):.3f}")
```

## Exercise 3

The negatives in the 3 × 4 matrix are −0.96 (row 1), −2.61 and −0.17 (row 2), −0.08 and −0.01 (row 3): **5 zeros**.

```python
relu = np.maximum(X @ W + bias, 0)
print(relu.round(2))
print("zeros:", (relu == 0).sum())
```
