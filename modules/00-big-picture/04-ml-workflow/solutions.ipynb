| Decision | Add a free returns label (cost £1.50) only to orders likely to be returned; or flag them for a quality check before shipping |
| Unit | One order, at the moment it is placed |
| Target | `returned` = 1 if any item on the invoice is returned within 30 days |
| Features | Known at order time: basket size, product categories, price, customer's past return rate, country, first order or not. **Not** allowed: delivery delay, courier damage report (happen later) |
| Metric | Precision of the flagged orders (how many really return), since each flag costs money |
| Baseline | "Flag orders over £200 from customers with a previous return" |
| Good enough | Flag half as many orders as the baseline for the same number of true returns |
| Risk | Return reasons we never record; the model learns product defects instead of customer behaviour |

## Exercise 3

```python
# --- data helper: works locally and on Colab -------------------------------
from pathlib import Path
import urllib.request

def data_path(name):
    """Return a local path to data/<name>, downloading from GitHub on Colab."""
    for parent in [Path.cwd(), *Path.cwd().parents]:
        candidate = parent / "data" / name
        if candidate.exists():
            return candidate
    url = f"https://raw.githubusercontent.com/cloudops-one/training-ai-ml/main/data/{name}"
    dest = Path(name)
    if not dest.exists():
        urllib.request.urlretrieve(url, dest)
    return dest

import pandas as pd
cust = pd.read_csv(data_path("shopwave_customers_small.csv"))

y = cust["spend_next_month"]
baseline = y.mean()
mae = (y - baseline).abs().mean()
print(f"predict £{baseline:.2f} for everyone  ->  mean absolute error £{mae:.2f}")
```
