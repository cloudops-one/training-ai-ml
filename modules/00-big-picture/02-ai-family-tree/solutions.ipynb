## Exercise 2

```python
import numpy as np, pandas as pd, time
from sklearn.tree import DecisionTreeClassifier, export_text
from sklearn.neural_network import MLPClassifier
from sklearn.preprocessing import StandardScaler
import warnings; warnings.filterwarnings("ignore")

rng = np.random.default_rng(1)
n = 10_000
big = pd.DataFrame({"total": rng.gamma(2, 150, n).round(),
                    "items": rng.integers(1, 15, n),
                    "account_days": rng.integers(1, 1000, n)})
big["returns"] = ((big["account_days"] > 90) | (big["total"] > 400)).astype(int)
Xb, yb = big[["total", "items", "account_days"]], big["returns"]

t0 = time.perf_counter()
tree = DecisionTreeClassifier(max_depth=3, random_state=0).fit(Xb, yb)
t_tree = time.perf_counter() - t0

Xs = StandardScaler().fit_transform(Xb)
t0 = time.perf_counter()
net = MLPClassifier(hidden_layer_sizes=(32, 32), max_iter=300, random_state=0).fit(Xs, yb)
t_net = time.perf_counter() - t0

print(f"tree: {t_tree:.3f}s, accuracy {(tree.predict(Xb) == yb).mean():.1%}")
print(f"net : {t_net:.3f}s, accuracy {(net.predict(Xs) == yb).mean():.1%}")
print(export_text(tree, feature_names=list(Xb.columns)))
```

The tree trains in a few milliseconds, recovers the hidden rule almost exactly, and you can read it. The network takes far longer, reaches similar accuracy, and its 1,000+ weights tell you nothing by inspection. On tabular data this is the usual outcome.
