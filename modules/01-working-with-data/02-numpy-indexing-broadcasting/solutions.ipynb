## Exercise 2

1. `(2, 3, 4)` with `(4,)`: from the right 4=4 ✓, then 3 vs nothing ✓, 2 vs nothing ✓ → **(2, 3, 4)**.
2. `(2, 3, 4)` with `(3, 1)`: 4 vs 1 ✓, 3 vs 3 ✓, 2 vs nothing ✓ → **(2, 3, 4)**.
3. `(2, 3)` with `(2,)`: 3 vs 2 ✗ → **error**. The `(2,)` array would need to be a column, `(2, 1)`.

```python
print((np.ones((2, 3, 4)) * np.arange(4)).shape)
print((np.ones((2, 3, 4)) * np.ones((3, 1))).shape)
try:
    np.ones((2, 3)) * np.ones((2,))
except ValueError as e:
    print("error:", e)
print("fixed:", (np.ones((2, 3)) * np.ones((2,))[:, None]).shape)
```

## Exercise 4

```python
# --- data helper: works locally and on Colab -------------------------------
from pathlib import Path
import urllib.request

def data_path(name):
    """Return a local path to data/<name>, downloading from GitHub on Colab."""
    for parent in [Path.cwd(), *Path.cwd().parents]:
        candidate = parent / "data" / name
        if candidate.exists():
            return candidate
    url = f"https://raw.githubusercontent.com/cloudops-one/training-ai-ml/main/data/{name}"
    dest = Path(name)
    if not dest.exists():
        urllib.request.urlretrieve(url, dest)
    return dest

import sys, pandas as pd
sys.path.insert(0, str(data_path("download.py").parent))
from download import ensure_online_retail

retail = pd.read_excel(ensure_online_retail(), sheet_name="Year 2010-2011")
qty_real = retail["Quantity"].to_numpy()
price_real = retail["Price"].to_numpy()
country = retail["Country"].to_numpy()
revenue_real = qty_real * price_real
print(revenue_real.shape)
```

```python
returns = qty_real < 0
print(f"return lines: {returns.sum():,}  ({returns.mean():.2%} of all lines)")
uk_returns = returns & (country == "United Kingdom")
print(f"value of UK returns: £{revenue_real[uk_returns].sum():,.0f}")
```
