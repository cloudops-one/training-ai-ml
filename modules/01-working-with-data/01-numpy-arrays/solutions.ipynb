## Exercise 2

`np.ones((3, 4)) * np.arange(4)` gives three identical rows `[0, 1, 2, 3]` (the 1-D array is applied to every row; that is broadcasting, lesson 01.2).

- `sum(axis=0)` adds down the columns: `[0, 3, 6, 9]`, shape `(4,)`.
- `sum(axis=1)` adds across the rows: `[6, 6, 6]`, shape `(3,)`.

```python
g = np.ones((3, 4)) * np.arange(4)
print(g)
print("axis=0:", g.sum(axis=0), g.sum(axis=0).shape)
print("axis=1:", g.sum(axis=1), g.sum(axis=1).shape)
```

## Exercise 4

```python
# --- data helper: works locally and on Colab -------------------------------
from pathlib import Path
import urllib.request

def data_path(name):
    """Return a local path to data/<name>, downloading from GitHub on Colab."""
    for parent in [Path.cwd(), *Path.cwd().parents]:
        candidate = parent / "data" / name
        if candidate.exists():
            return candidate
    url = f"https://raw.githubusercontent.com/cloudops-one/training-ai-ml/main/data/{name}"
    dest = Path(name)
    if not dest.exists():
        urllib.request.urlretrieve(url, dest)
    return dest

import sys, pandas as pd
sys.path.insert(0, str(data_path("download.py").parent))
from download import ensure_online_retail

retail = pd.read_excel(ensure_online_retail(), sheet_name="Year 2010-2011")
qty_real = retail["Quantity"].to_numpy()
price_real = retail["Price"].to_numpy()
print(qty_real.shape, qty_real.dtype, "|", price_real.shape, price_real.dtype)
```

```python
revenue_real = qty_real * price_real
big = revenue_real > 1000           # a boolean array, True where the condition holds
print("lines above £1,000:", big.sum())                       # True counts as 1
print(f"fraction: {big.mean():.4%}")                           # mean of booleans = fraction True
```

`big` is an array of `True`/`False`. Summing it counts the `True`s; taking its mean gives the fraction. This trick is used constantly; lesson 01.2 makes it official.
