```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style="whitegrid")
```

```python
# --- data helper: works locally and on Colab -------------------------------
from pathlib import Path
import urllib.request

def data_path(name):
    """Return a local path to data/<name>, downloading from GitHub on Colab."""
    for parent in [Path.cwd(), *Path.cwd().parents]:
        candidate = parent / "data" / name
        if candidate.exists():
            return candidate
    url = f"https://raw.githubusercontent.com/cloudops-one/training-ai-ml/main/data/{name}"
    dest = Path(name)
    if not dest.exists():
        urllib.request.urlretrieve(url, dest)
    return dest

import sys, pandas as pd
sys.path.insert(0, str(data_path("download.py").parent))
from download import ensure_online_retail

raw = pd.read_excel(ensure_online_retail(), sheet_name="Year 2010-2011")
print(raw.shape)
raw.head()
```

```python
def clean_retail(df):
    """The standard ShopWave cleaning used from lesson 01.5 onwards."""
    out = df.copy()
    out = out[~out["Invoice"].astype(str).str.startswith("C")]   # drop cancellations
    out = out[(out["Quantity"] > 0) & (out["Price"] > 0)]         # drop returns / free items
    out = out.dropna(subset=["Customer ID"])                       # drop guest checkouts
    out = out.drop_duplicates()
    out["Customer ID"] = out["Customer ID"].astype(int)
    out["Revenue"] = out["Quantity"] * out["Price"]
    return out

sales = clean_retail(raw)
print(f"{len(raw):,} rows → {len(sales):,} after cleaning")
```

```python
def overview(df, target=None):
    """Checklist steps 1-5 in one call. Prints; returns nothing."""
    print(f"1. SHAPE      {df.shape[0]:,} rows × {df.shape[1]} columns")
    print("2. TYPES")
    types = pd.DataFrame({"dtype": df.dtypes.astype(str), "distinct": df.nunique()})
    print(types.to_string())
    miss = df.isna().mean().mul(100).round(1)
    print("3. MISSING %  " + (miss[miss > 0].to_dict().__repr__() if (miss > 0).any() else "none"))
    print(f"4. DUPLICATES {df.duplicated().sum():,} exact")
    if target is not None:
        t = df[target]
        if t.nunique() <= 10:
            print(f"5. TARGET     {t.value_counts(normalize=True).round(3).to_dict()}")
        else:
            print(f"5. TARGET     mean {t.mean():.1f}, median {t.median():.1f}, min {t.min():.1f}, max {t.max():.1f}, zeros {(t == 0).mean():.0%}")
```

## Exercise 1

```python
overview(raw)
```

Findings: 541,910 rows, one row per *invoice line*, not per order or customer. `Customer ID` 25% missing (guest checkouts) and stored as float for that reason. 5,268 exact duplicates. `Invoice` and `StockCode` are IDs with thousands of distinct values, not features. `Quantity` and `Price` have negative and zero values (returns, adjustments).

Steps 5, 7 and 8 (target, feature-vs-target, leakage) do not apply: a transaction table has no target. You first decide the unit of analysis (customer) and aggregate; then those steps make sense.

```python
cutoff = pd.Timestamp("2011-09-01")
past = sales[sales["InvoiceDate"] < cutoff]
future = sales[(sales["InvoiceDate"] >= cutoff) & (sales["InvoiceDate"] < "2011-12-01")]

cust = past.groupby("Customer ID").agg(
    orders=("Invoice", "nunique"), spend=("Revenue", "sum"),
    products=("StockCode", "nunique"), items=("Quantity", "sum"),
    first_order=("InvoiceDate", "min"), last_order=("InvoiceDate", "max"),
    country=("Country", "first"))
cust["avg_order"] = cust["spend"] / cust["orders"]
cust["tenure_days"] = (cutoff - cust["first_order"]).dt.days
cust["days_since_last"] = (cutoff - cust["last_order"]).dt.days

# target: what each customer spent in the next quarter (0 if they did not come back)
cust["spend_next_quarter"] = future.groupby("Customer ID")["Revenue"].sum().reindex(cust.index).fillna(0)
cust["came_back"] = (cust["spend_next_quarter"] > 0).astype(int)

# a column "helpfully" added by the export: recency measured at export time (Dec 2011)
last_ever = sales.groupby("Customer ID")["InvoiceDate"].max()
cust["recency_days"] = (sales["InvoiceDate"].max() - last_ever).dt.days.reindex(cust.index)

cust = cust.drop(columns=["first_order", "last_order"])
print(cust.shape)
cust.head()
```

## Exercise 2

```python
counts = cust["came_back"].value_counts()
print(f"imbalance 1 : {counts.max()/counts.min():.2f}")
feats = ["orders", "spend", "products", "items", "avg_order", "tenure_days", "days_since_last"]
med = cust.groupby("came_back")[feats].median()
ratio = (med.loc[1] / med.loc[0]).round(2)
print(ratio.sort_values())
```

`days_since_last` (returners' median is a fraction of non-returners') and `orders` separate best. `avg_order` barely differs: basket size does not predict coming back; frequency and recency do.

## Exercise 3

```python
span = past.groupby("Customer ID")["InvoiceDate"].agg(["min", "max"])
gap = (span["max"] - span["min"]).dt.days / (cust["orders"] - 1)
cust["avg_days_between_orders"] = gap.replace([np.inf, -np.inf], np.nan)
print(f"missing rate: {cust['avg_days_between_orders'].isna().mean():.1%}")
print("came_back rate when missing:", cust.loc[cust["avg_days_between_orders"].isna(), "came_back"].mean().round(3))
print("came_back rate when present:", cust.loc[cust["avg_days_between_orders"].notna(), "came_back"].mean().round(3))
```

Missing for every customer with exactly one order (division by zero). Those one-time buyers come back far less often than repeat buyers, so the *missingness itself* predicts the target. Keep a flag column `one_order = orders == 1` rather than silently imputing.

## Exercise 4

```python
num = cust.drop(columns=["country", "came_back", "spend_next_quarter", "recency_days"]).corr().abs()
pairs = (num.where(np.triu(np.ones(num.shape), k=1).astype(bool))
            .stack().sort_values(ascending=False))
pairs[pairs > 0.6].round(2)
```

`spend`–`items` (0.92) is the strongest; `orders`–`products` and the other "how much they buy" pairs follow. Keep `orders` (cleanest meaning, used in RFM) and `spend` (monetary); drop `items`, and `products` unless you have a reason to keep it. A model can cope with correlated features, but you cannot *explain* a model that has four versions of "buys a lot".
