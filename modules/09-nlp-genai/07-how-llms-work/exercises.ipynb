## Exercise 1: your own bigram model

Add the sentences *"the lamp is big"* and *"the mug is blue"* to the three from the lesson. By hand: what is P(*mug* | *the*) now? P(*red* | *is*)? What does greedy decoding write? Then check in code.

```python
from collections import Counter, defaultdict

sentences = ["the mug is red", "the mug is big", "the lamp is red"]
counts = defaultdict(Counter)                 # counts[word][next_word]
for s in sentences:
    words = s.split()
    for a, b in zip(words, words[1:]):
        counts[a][b] += 1

for word, nxt in counts.items():
    total = sum(nxt.values())
    print(f"after '{word}':", {w: f"{c}/{total}" for w, c in nxt.items()})
```

## Exercise 2: extreme temperatures

By hand (a calculator is fine), compute the softmax of the scores [3.0, 2.5, 1.5, −1.0] at T = 0.1 and T = 5. What happens to the probabilities as T → 0 and T → ∞? Check with the `softmax` function.

```python
import numpy as np

candidates = ["order", "purchase", "patience", "banana"]
scores = np.array([3.0, 2.5, 1.5, -1.0])       # the model's raw scores (logits)

def softmax(z, temperature=1.0):
    e = np.exp(z / temperature)
    return e / e.sum()

for T in [0.5, 1.0, 2.0]:
    probs = softmax(scores, T)
    print(f"T={T}: " + "  ".join(f"{w} {p:.3f}" for w, p in zip(candidates, probs)))
```
