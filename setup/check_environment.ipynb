```python
import sys, platform
print("Python", sys.version.split()[0], "on", platform.system(), platform.machine())
major, minor = sys.version_info[:2]
if (major, minor) == (3, 12):
    print("Python version: OK (3.12 is what the course is tested on)")
elif (major, minor) in [(3, 10), (3, 11)]:
    print("Python version: should work, but 3.12 is recommended")
else:
    print("Python version: WARNING, some libraries (TensorFlow, PyCaret) may not install. Use 3.12.")
```

```python
import importlib, os
os.environ["MLFLOW_DISABLE_AGENT_HINT"] = "1"   # silence an informational message from mlflow

CORE = ["numpy", "pandas", "matplotlib", "seaborn", "scipy", "sklearn", "statsmodels",
        "xgboost", "lightgbm", "imblearn", "shap", "mlxtend", "optuna", "joblib", "mlflow", "streamlit", "fastapi"]
DEEP = ["tensorflow", "keras", "torch"]
GENAI = ["transformers", "sentence_transformers", "anthropic", "chromadb"]

def check(names, label):
    missing = []
    print(f"\n{label}")
    for n in names:
        try:
            m = importlib.import_module(n)
            print(f"  ok   {n:<22} {getattr(m, '__version__', '')}")
        except Exception as e:
            print(f"  --   {n:<22} missing")
            missing.append(n)
    return missing

missing_core = check(CORE, "Core (needed from Module 00)")
missing_deep = check(DEEP, "Deep learning (needed from Module 08)")
missing_genai = check(GENAI, "Generative AI (needed from Module 09)")
```

```python
import numpy as np
from sklearn.linear_model import LinearRegression
x = np.array([[2], [4], [6], [8], [10]]); y = np.array([50, 80, 90, 130, 150])
m = LinearRegression().fit(x, y)
assert abs(m.coef_[0] - 12.5) < 1e-9 and abs(m.intercept_ - 25) < 1e-9
print("scikit-learn smoke test: OK (slope 12.5, intercept 25)")
```
